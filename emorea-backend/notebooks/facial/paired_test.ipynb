```python
import numpy as np
from scipy.stats import ttest_rel

def delta_uar_with_significance(uar_cnn, uar_mvit):
    uar_cnn = np.array(uar_cnn)
    print("Mean UAR CNN:", np.mean(uar_cnn), "Std UAR CNN:", np.std(uar_cnn, ddof=1))
    uar_mvit = np.array(uar_mvit)
    print("Mean UAR MViT:", np.mean(uar_mvit), "Std UAR MViT:", np.std(uar_mvit, ddof=1))

    delta = uar_mvit - uar_cnn

    delta_mean = np.mean(delta)
    delta_std = np.std(delta, ddof=1)

    _, p = ttest_rel(uar_mvit, uar_cnn)

    if p < 0.01:
        sig = "**"
    elif p < 0.05:
        sig = "*"
    else:
        sig = "n.s."

    print(f"ΔUAR = {delta_mean:.3f} ± {delta_std:.3f}, p = {p:.4f}, signal: {sig}")

    return delta_mean, delta_std, p, sig
```

```python
delta_uar_with_significance([0.51, 0.49, 0.49], [0.54, 0.56, 0.56])
```
